```python
from sympy import *
```

```python
# For gaussians, this seems to be a good reference for dealing with combination of damping
# functions and ewald: Efficient Handling of Gaussian Charge Distributions: An Application to Polarizable Molecular Models
# I am going to first validate the direct damping factors for Slater functions (which I know) so that I can
# then apply this to the case of erfc which is the case we need to consider for ewald.

x = symbols('x')
```

```python
m3 = 1
m5 = 3
m7 = 15
m9 = 105
m11 = 945
m13 = 10395
```

```python
C0 = -1/x * erf(x)
C1 = (-1/x) * diff(C0, x)
C2 = (-1/x) * diff(C1, x)
C3 = (-1/x) * diff(C2, x)
C4 = (-1/x) * diff(C3, x)
C5 = (-1/x) * diff(C4, x)
C6 = (-1/x) * diff(C5, x)
display(expand(C0))
display(expand(C1))
display(expand(C2))
display(expand(C3))
display(expand(C4))
display(expand(C5))
display(expand(C6))

#C0 = 1/x * erfc(x)
#C1 = (-1/x) * diff(C0, x)
#C2 = (-1/x) * diff(C1, x)
#C3 = (-1/x) * diff(C2, x)
#C4 = (-1/x) * diff(C3, x)
#C5 = (-1/x) * diff(C4, x)
#C6 = (-1/x) * diff(C5, x)
#display(expand(C0))
#display(expand(C1))
#display(expand(C2))
#display(expand(C3))
#display(expand(C4))
#display(expand(C5))
#display(expand(C6))
```

```python
P3  = x / m3
P5  = (3*x + 2*x**3) / m5
P7  = (15*x + 10*x**3 + 4*x**5) / m7
P9  = (8*x**7 + 28*x**5 + 70*x**3 + 105*x) / m9
P11 = (16*x**9 + 72*x**7 + 252*x**5 +630*x**3 + 945*x) / m11
P13 = (32*x**11 + 176*x**9 + 792*x**7 + 2772*x**5 + 6930*x**3 + 10395*x) / m13

f1 = -erf(x)
f3 = -erf(x) + 2/sqrt(pi) * P3 * exp(-x**2)
f5 = -erf(x) + 2/sqrt(pi) * P5 * exp(-x**2)
f7 = -erf(x) + 2/sqrt(pi) * P7 * exp(-x**2)
f9 = -erf(x) + 2/sqrt(pi) * P9 * exp(-x**2)
f11 = -erf(x) + 2/sqrt(pi) * P11 * exp(-x**2)
f13 = -erf(x) + 2/sqrt(pi) * P13 * exp(-x**2)
display(C1.equals(m3 * f3 / x**3))
display(C2.equals(m5 * f5 / x**5))
display(C3.equals(m7 * f7 / x**7))
display(C4.equals(m9 * f9 / x**9))
display(C5.equals(m11 * f11 / x**11))
display(C6.equals(m13 * f13 / x**13))
```
